```python
import requests
import pandas as pd
```

```python
data = requests.get("https://data.police.uk/api/crimes-no-location?category=all-crime&date=2024-03&force=leicestershire")
data = data.json()
data = pd.DataFrame(data)
data
```

```python
data
```

```python
categories = []
for status in data['outcome_status']:
    categories.append(status['category'])
    
print("categories: ", categories)
data['status'] = categories
```

```python
stakeholder_data = data[
    (data['category']=='public-order') 
    & 
    (data['status']=="Unable to prosecute suspect")
    |
    (data['status'] == "Court result unavailable")
    ]
stakeholder_data
```
