```python
import pandas as pd
import requests
```

```python
crime_severity_data = pd.read_csv(r"data\crime_severity.csv")
crime_severity_data.head()
```

```python
crime_severity_data =crime_severity_data[
    (crime_severity_data['category']=='public-order') 
     &(crime_severity_data['severity']>=2)
    ] 
crime_severity_data.head()
```

```python
combined_data = requests.get("https://data.police.uk/api/crimes-no-location?category=all-crime&date=2024-03&force=leicestershire")
combined_data = combined_data.json()
combined_data= pd.DataFrame(combined_data)

categories = []
for status in combined_data['outcome_status']:
    categories.append(status['category'])
    
print("categories: ", categories)
combined_data['status'] = categories

combined_data = combined_data[
    (combined_data['category']=='public-order') 
    & 
    (combined_data['status']=="Unable to prosecute suspect")
    |
    (combined_data['status'] == "Court result unavailable")
    ]
combined_data
```

```python
combined_data.set_index('id')
crime_severity_data.set_index('id')
```

```python
data = pd.merge(combined_data, crime_severity_data, on='id')
data
```

```python
severity = []
for crime in data['severity']:
    if crime == 2:
        severity.append('low')
    elif crime == 3:
        severity.append('medium')
    elif crime == 4:
        severity.append('high')
data['sev-cat']=severity
data.to_csv("task_4_data.csv")
```
